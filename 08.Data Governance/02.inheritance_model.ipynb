
El equipo de ventas se ha enterado de la gente de bi tenía un schema secreto con tablas muy interesantes de salesforce que podrían ser el futuro y cambiarlo todo. Exigen, demandan que les des acceso. 

```python
%sql

-- Let's create the tables

create schema if not exists workspace.golden_bi;

create table if not exists workspace.golden_bi.clicks (
    id int,
    timestamp string,
    os string
);
create table if not exists workspace.golden_bi.pushes (
     id int,
    timestamp string,
    os string
)
```


El path estándar, es dar acceso a estas tablas

```python
%sql
grant use catalog on catalog workspace to marketing;
grant use schema on schema workspace.golden_bi to marketing;
grant select on table workspace.golden_bi.clicks to marketing;
grant select on table workspace.golden_bi.pushes to marketing;
```


El equipo de marketing está hecho una furia, resulta que había más tablas, que todas las tablas que necesitan estan en golden_bi y que el director ha decidido que para evitar esto, pueden leer todas las tablas de bi.

```python
%sql 

create table if not exists workspace.golden_bi.email (
     id int,
    timestamp string,
    os string
)
```

```python
%sql 

-- Si intentaramos leer no nos dejaría, pero aprovechamos el modelo de herencia de unity catalog 
grant select on schema workspace.golden_bi to marketing;
```
