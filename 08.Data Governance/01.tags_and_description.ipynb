```python
%sql

-- Vamos a usar tags para categorizar bien nuestros productos, pongamos que tenemos unas tablas de usuarios en silver, de la cual dependen tablas de sales y marketing
create schema workspace.silver;
```

```python
%sql
create table workspace.silver.users (
    id int, 
    name string, 
    registration_ts timestamp 
)
```

```python
%sql
SET TAG ON table workspace.silver.users `SALES`;
SET TAG ON table workspace.silver.users `MARKETING`;

```

```python
%sql
-- To show these new schemas we can use the  information_schema
select * from INFORMATION_SCHEMA.TABLE_TAGS
```

As we can see tags can use tag_name and tag_value, useful tags are owner, projects, area... These can land in a unified view where resources can be organized in a simple way.

```python
%sql
CREATE GOVERNED TAG visibility VALUES ('all', 'admin', 'data');
```

```python
%sql
ALTER GOVERNED TAG visibility SET VALUES ('all', 'admin', 'data', 'none');
```
